what is data visualization ?
it is  the process of representing data in a grafical or
visuals format such as charts , graphs or maps so that people
can easily undestand ppatterns, trens and insights

why visualization is important?
* makes data easy to undastand
* helps in quick desision making
* identifies patters and trends 
*detects outliers or errors 
*improves communication of insigts

Univariate analysis :
it means analyzing only one variable at a time to check the quality of the data.

purpose:
*u
 undestand distribution 
* find patterns
* detect outliers 
ex: histogram , box plot , KDE plot , Violin plot 

ex: analyse salary distribution  in a company are most emploee low paid or high paid

Bivarite analysis:
it is means analyzing the relationship netween two variables.
purpose:
* undestand relationship
* identify corelation
* compare variable
ex: scatter plot , bar plot , line plot

ex: analyze advertising spend vs sales
does increasing ads increase sales ?

multivariate analysis:
it means analyzing relationship between more then two variables.

purpose:
* undastand complax relationships 
* identify pattern across multiple variables 
* perform deep data analysis (EDA)
(EDA) : pair plot , heatmap, facet grid , bubble plot

ex: analyze sales based on region , time and product category 
-> which combination gives highest profit?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

histogram:
* shows distribution of data 
* helps to detect skewness and spread

```python
# import numpy as np
plt.figure(figsize = (5,5))
plt.hist(df["total_bill"])
plt.title("histogram")
plt.xlabel("total_bill")
plt.ylabel("frequency")
plt.xticks()
plt.yticks(np.arrange(10,80,10))
plt.show()
```

KDE plot: (kernal density estimation plot)
* smooth version of histogram 
* better for undastanding data density

```python
plt.figure(facecolor='black')
sns.kdeplot(df['total_bill'],fill = True)
plt.title("kdeplot")
plt.xlabel("total_bill")
plt.ylabel("frequency")
plt.show()
```

box plot:
* identifies outliers
* shows median , quartiles, spread

```python
plt.figure()
sns.boxplot(df['total_bill'],orient="h")
plt.title("box plot")
plt.xlabel("total_bill")
# plt.ylabel("frequency")
plt.show()
```

Violin plot :
*combines bow plot + distribution 
* shows data shape clearly

```python
plt.figure()
sns.violinplot(x=df["total_bill"])
plt.title("violin plot")
plt.xlabel("total_bill")
# plt.ylabel("frequency")
plt.show()
```

Bar plot :
* compares values across categories
* easy to understand differences

```python
plt.figure()
sns.barplot(x="total_bill",y="day",data=df,orient="h")
plt.title("bar plot")
plt.show()
```

```python
pie chart:
*shows proportion / percentange
* good for quick insights 
note : use only for small num of categories 
```

```python
plt.figure()
df['day'].value_counts().plot.pie(autopct='%1.2f%%')
# autopct='%1.2f%%' is use to show inside value
plt.title("pie chart")
plt.show()
```

line plot:
*shows trends over time 
* helps to identify increases,decreades,seasonality
ex: sales growth month wise

```python
plt.plot(df["total_bill"])
plt.title("Line plot")
plt.show()
```

count plot:
* shows frequency of categories
*useful for categorical analysis
ex: number of customers per city

```python
sns.countplot(x="day",data=df)
plt.title("count plot") 
plt.show()
```

scatter plot:
*shows relationship between two numerical variables 
* helps detect corelation and patterns
ex: Advertising Vs Sales

```python
sns.scatterplot(x="total_bill",y="tip",data=df)
plt.title("scatter plot")
plt.show()
```

```python
there are three type of co relation:
*positive corelation
*negetive corelation 
*no corelation
```

```python
bubble sort:
*add third variable (size)
* shows more information in one chart

ex: sales vs profit with market size
```

```python
plt.scatter(df['total_bill'],df['tip'],s=df['size']*10)
plt.tital("bubble plot")
plt.show()
```

```python
pair plot:
* shows relationship between all variables
* useful in EDA(explotoratory data analysis)

ex: feature relationship in ML
```

```python
sns.pairplot(df)
plt.show()
```

```python
Heatmap : 
* shows corelation between variables
* helps in feature selection 
ex: identity highly corelated feature 
```

```python
corr = df.corr(numeric_only = True)
corr.round(2)
```

```python
sns.heatmap(corr,annot = True)
plt.show()
```
