```python
# create a function to print prime number
# with given range from user
# using function without input without return
```

```python
def prime():
    sv = int(input("enter a start value"))
    sp = int(input("enter a stop value"))
    for i in range (sv,sp+1,1):
        for j in range(2,i,1):
            if(i%j==0):
                break
        else:
            print(i,end=" ")

prime()
    
```

```python
def prime(sv,sp):
    pl= []
    sv = int(input("enter a start value"))
    sp = int(input("enter a stop value"))
    for i in range (sv,sp+1,1):
        for j in range(2,i,1):
            if(i%j==0):
                break
        else:
            pl.append(i)

    return pl
prime(50,100)
```

```python
lambda function
a lambda function is a small,anonymous(nameless)
function in python

*it defined using keyword lambda insted of def 
*it can take any number of argument but must containe only one expression
*expression is automatically returned(no need to use return)
*single line function 

synteax:
lambda arguments: expression
```

```python
sq=lambda x : x*x
sq(4)
```

```python
add = lambda a,b : a+b
add(5,7)
```

```python
why to use lambda?
*useful for short,simple operation
*saves code(one-liner)
*often used with map(),filter(),reduce()
or as inline function

when not use lambda?
*for complex logic(use def insted)
*when readility is more important then shortness.
```

```python
create function to calucalte factorial of given number 
using without input with return
```

```python
def fact():
    n = int(input("enter a num"))
    f = 1
    for i in range (n,0,-1):
        f-f*i

    return  n,f

r = fact()
# the factorial of 5 is 120
```

```python
v,r = fact()
```

```python
print(f"the factorial of {v} is {r}")
```
