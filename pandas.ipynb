```python
ins = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\insurance.csv")
ins
```

```python
# Group by region and sum the charges 
ins.groupby('region')['expenses'].sum()
```

```python
# region and sum the charges  by pivot
pd.pivot_table(data=ins,values='expenses',index='region', aggfunc='sum')
```

```python
# Group by region and sum the charges  in ascending order
ins.groupby('region')['expenses'].sum().sort_values()
```

```python
# Group by region and sum the charges  in decending order
ins.groupby('region')['expenses'].sum().sort_values(ascending=False)
```

```python
# gender wise avg bmi and expenses by group by
ins.groupby('sex')[['bmi', 'expenses']].mean().round()
```
